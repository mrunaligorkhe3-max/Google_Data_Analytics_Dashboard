## 15. Handling Missing Values

```python
#fill missing Profit with mean & median.
```

```python
Google["Profit"] = Google["Profit"].fillna(Google["Profit"].mean())
Google["Profit"].isnull().sum()
```

```python
Google["Profit"] = Google["Profit"].fillna(Google["Profit"].median())
Google["Profit"].isnull().sum()
```

```python
# fill Total_Sales with mode.
```

```python
Google["Total_Sales"] = Google["Total_Sales"].fillna(Google["Total_Sales"].mode()[0])
Google["Total_Sales"].isnull().sum()
```

```python
# fill forward.
```

```python
Google.ffill(inplace=True)
```

```python
# backforward fill.
```

```python
Google.bfill(inplace=True)
```

## 16. Checking Duplicates

```python
# checking duplicate rows.
```

```python
Google.duplicated().sum()
```

## 17. Removing Duplicates

```python
Google.drop_duplicates(inplace=True)
```

```python
#remove missing rows.
```

```python
Google.dropna(inplace=True)
```

## 18. Rename Columns

```python
# renaming columns to improve readability and consistency in the dataset.
```

```python
Google.rename(columns={
    "Product_Category": "Product",
    "Unit_Price": "Price",
    "Customer_Rating": "Rating",
    "Total_Sales": "Sales"
}, inplace=True)
```

# Section J : Visualization → How can I show the results?

## 61. BAR CHART — Sales by Product Category

```python
colors = ["#264653", "#2A9D8F", "#E9C46A", "#F4A261", "#E76F51"]

Google.groupby("Product")["Sales"].sum().plot(
    kind="barh",
    figsize=(6,4),
    color=colors
)

plt.title("Total Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 62. PIE CHART — Orders by Product Category

```python
colors = ["#FF6B6B", "#4ECDC4", "#45B7D1", "#96CEB4", "#FFEAA7"]

Google["Product"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(6,4),
    colors=colors
)

plt.title("Order Distribution by Product Category")
plt.ylabel("")
plt.show()
```

## 63. HISTOGRAM — Customer Age Distribution

```python
colors = "#AB83A1"

Google["Age"].plot(
    kind="hist",
    bins=10,
    figsize=(8,4),
    color=colors
)

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

## 64. LINE GRAPH — Monthly Sales

```python
colors = "#901E3E"

Google["Order_Date"] = pd.to_datetime(Google["Order_Date"])

monthly_sales = Google.groupby(
    Google["Order_Date"].dt.to_period("M")
)["Sales"].sum()

monthly_sales.plot(
    kind="line",
    figsize=(10,5),
    marker="o",
    color=colors
)

plt.title("Monthly Total Sales")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 65. BAR CHART — Orders by Marketing Channel

```python
colors = ["#6A0572", "#AB83A1", "#D4A5A5", "#F0D9DA", "#F7F3F5", "#3C096C"]

Google["Marketing_Channel"].value_counts().plot(
    kind="bar",
    figsize=(8,5),
    color=colors
)

plt.title("Number of Orders by Marketing Channel")
plt.xlabel("Marketing Channel")
plt.ylabel("Order Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 66. PIE CHART — Payment Method Distribution
